```python
# 1. CSV 파일을 읽어 data 리스트([{}, {}, {}, ...])를 생성하시오.

import csv
```

```python
# 파일 같은 자원을 불러올 땐 메모리 로드 후 자원 해지가 필요하다
with open("s.csv", "r", encoding="utf-8") as f: #("파일 이름", "읽기 전용", "한글 옵션")
    reader = csv.DictReader(f) #csv.DictReader = interable(반복 가능함)
    print(reader, type(reader))
    data = list(reader)

data
```

```python
# 2. 학생들의 이름/국어/영어 리스트 각각 저장 후 출력

#변수 설정
names = []
kors = []
engs = []

#for문 사용
# for row in data:
#     names.append(row['이름'])
#     korean.append(row['국어'])
#     english.append(row['영어'])

#컴프리헨션 사용
names = [r["이름"] for r in data]
kors = [r["국어"] for r in data]
engs = [r["영어"] for r in data]

print(names)
print(kors)
print(engs)
```

```python
# 3. zip()을 이용하여 학생 이름과 국어 점수를 [(), (), ..] 형태로 저장 후 출력

name_kor = list(zip(names, kors))
name_eng = list(zip(names, engs))
print(name_kor)
print(name_eng)
```

```python
# 4~5. dict(zip())을 이용하여 
# 다음과 같은 딕셔너리(2개 name_kor_dict, name_eng_dict)를 저장 후 출력

name_kor_list = list(zip(names, kors)) #반복자 커스 이동 처음 → 끝
name_eng_list = list(zip(names, engs))
name_kor_dict = dict(zip(names, kors))
name_eng_dict = dict(zip(names, engs))

print(name_kor_list)
print(name_eng_list)
print(name_kor_dict)
print(name_eng_dict)
```

```python
# 7. 국어/영어 점수가 높은 순으로 정렬하시오. (sortede/lambda 사용)

#국어
sorted_k = sorted(name_kor_list, key=lambda x : int(x[1]), reverse = True)

#영어
sorted_e = sorted(name_eng_list, key=lambda x : int(x[1]), reverse = True)

print(sorted_k)
print(sorted_e)
```

```python
# 8. 국어와 영어의 평균을 계산하여 다음과 같은 자료구조를 생성하시오.
# [
# ('홍길동',87.5),
# ('이순신',89.0),
# ...
# ]

name_kor_eng = zip(names, kors, engs) #먼저 데이터 묶어주기
names_avg_list = [(n, (int(k)+int(e))/2) for n, k, e in name_kor_eng]
print(names_avg_list)
```

```python
# 9. (람다 사용하여) 문제 8의 결과에서 평균이 90점 이상인 학생만 출력하시오
```

```python
# 10. 문제 8의 결과에서 평균이 90점 이상인 학생만 출력하시오
# filter(), 람다 사용

list(filter(lambda x: x[1] >=90 , names_avg_list))
```
