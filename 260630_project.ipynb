```python
import csv

with open("score.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    data = list(reader)

print(data)
```

```python
# 1-1. 결측치 확인
# 나이/점수에서 숫자가 아닌 형태의 자료를 결측치로 처리

missing_counts = {'이름':0,'나이':0,'국어':0,'영어':0}

for row in data:
    for a, b in row.items():
        if a == '이름':
            continue

        try:
            int(b) #숫자로 바꿔보기
        except ValueError:
            missing_counts[a] += 1

print(missing_counts)
```

```python
# 1-2. 결측치 처리 (국어/영어 점수)
# 결측치 = 숫자 형태가 아닌 걸 결측치로 처리함 / 결측치 = 0 상정

for row in data:
    for a, b in row.items():
        if a == "이름" or a == "나이":
            continue

        try:
            int(b)
        except ValueError:
            row[a] = 0

for row in data:
    print(row)
```
