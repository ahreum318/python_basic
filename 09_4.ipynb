```python
# enumerate() = 인덱스와 요소 묶기

names = ["홍길동", "이순신", "강감찬"]

for i, name in enumerate(names):
    print(i, name)

for i, name in enumerate(names):
    print(f"({i+1})", name)
```

```python
# reverse() 역순반복

nums = [1, 2, 3]
print(list(reversed(nums)))
```

```python
# any() 하나라도 True인지 검사
nums = [0, 0, 3]
print(any(nums))

# all() 모두 True인지 검사
nums = ["", "2", "3"]
print(all(nums))

nums = [{}, {}, {}]
print(all(nums))
```

```python
# zip() 여러 개의 시퀀스 객체를 같은 위치(인덱스)의 값끼리 묶어줌

names = ["홍길동", "이순신", "김유신"]
scores = [90, 80, 100]
gender = ["F", "M", "M"]
result = zip(names, scores, gender)
print(list(result))
```

```python
names = ["홍길동", "이순신", "김유신"]
scores = [90, 80, 100]
result = dict(zip(names, scores))

print(result)
```
