```python
# random.random() 함수만으로 로또 만들기
import random

# 1. 공을 45개 만들어야 됨 -> 변수 45개 생성
balls = list(range(1, 46))
print(balls)
```

```python
# 2. 볼 섞기 (ball -> shuffle)
for i in range(0, 1000000): #1,000,000번 반복
    temp = balls[0]
    r_num = int(random.random() * len(balls)) #0~44까지의 정수
    # r_num = math.floor(random.ramdom() * len(balls))
    balls[0] = balls[r_num]
    balls[r_num] = temp

print(balls[:6])
```

```python
for i in range(10):
    lotto = random.sample(range(1, 46), 6)
    lotto.sort()
    print(f"{i}번째 로또 번호: {lotto}")
```

```python
def my_lotto_creat(a):
    result = []
    for i in range(a):
        lotto = random.sample(range(1, 46), 6)
        lotto.sort()
        #print(f"{i}번째 로또 번호: {lotto}")
        result.append(tuple(lotto))
    return result
```

```python
result = my_lotto_creat(3)
print(result)
```

```python
# 주사위 큰 숫자 이기는 거 고르기

def dice_cast():
    d1 = random.randint(1, 6)
    d2 = random.randint(1, 6)
    return d1 + d2

#p1_dice = {"name": "홍길동", "dice_value": 0}
#p1_dice["dice_value"] = dice_cast()
p1_dice = dice_cast()
p2_dice = dice_cast()

if p1_dice == p2_dice:
    print(f"비겼다. p1: {p1_dice}, p2: {p2_dice}")
elif p1_dice > p2_dice:
    print(f"p1_dice Win!! p1: {p1_dice}, p2: {p2_dice}")
else:
    print(f"p2_dice Win!! p1: {p1_dice}, p2: {p2_dice}")


    
```
