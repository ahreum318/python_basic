```python
with open("students.txt", "r", encoding="utf-8") as f:
    # 한 줄씩 읽기.
    for line in f: 
        print(line) # 라인 끝의 줄바꿈이 문자 포함되어 두번 줄바꿈이 일어난다(print함수도 줄바꿈이 기본값)
        print(type(line))
```

```python
with open("students.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip()) # line에 양쪽 공백제거(줄바꿈포함)하여 print함수의 기본 줄바꿈만으로 출력
```

```python
# 한 라인을 구분자","를 이용하여 두개의 문자열 값으로 저장
names = []
scores = []

with open("students.txt", "r", encoding="utf-8") as f:
    for line in f:
        name, score = line.strip().split(",")
        names.append(name)
        scores.append(score)

print(names)
print(scores)
```

```python
# 딕셔너리 만들기
students = {}

with open("students.txt", "r", encoding="utf-8") as f:
    for line in f:
        name, score = line.strip().split(",")
        students[name] = int(score)

print(students)
```

```python
s = [] # [(), (), ()]

with open("students.txt", "r", encoding="utf-8") as f:
    for line in f:
        a = line.strip().split(",")
        s.append(tuple(a))

print(s)
```

```python
import csv

#open()은 내장 함수
with open("students.csv", "r", encoding="utf-8") as f:
    #reader() : csv모듈 안에 있는 함수
    reader = csv.reader(f)
    print(type(reader))

    for row in reader:
        print(row)
        print(type(row))
```

```python
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)

    next(reader) #제일 처음 바로 다음 거부터 읽어라
    print(type(reader))

    for row in reader:
        print(row)
```

```python
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for row in reader:
        print(row)
```

```python
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    data = list(reader)

print(data)

data[2]["영어"] = 80
print(data[2])
```

```python
# students_result.csv 파일을 쓰기(w) 모드로 열기
# newline="" : 빈 줄이 추가되는 현상 방지
# encoding="utf-8-sig" : 한글이 엑셀에서 깨지지 않도록 저장

with open("students_result.csv", "w", newline="", encoding="utf-8-sig") as f:

    # data[0].keys() → 첫 번째 딕셔너리의 키(컬럼명) 사용    
    writer = csv.DictWriter(f, fieldnames=data[0].keys())

    #여기서 수정 작업



    # 헤더(컬럼명) 저장
    writer.writeheader()   

    # data 리스트에 있는 모든 딕셔너리를 한 번에 저장
    writer.writerows(data)  

print("CSV 저장 완료")
```
