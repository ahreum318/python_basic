```python
even = []

for i in list_nums:
    if i%2 == 0:
        even.append(i)

print(even)
```

```python
even2 = [i for i in list_nums if i%2 == 0]
print(even2)

even3 = ([i*10 for i in list_nums if i<=5])
print(even3)
#위 코드와 같은 코드
```
