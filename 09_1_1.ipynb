```python
def add3(a, b):
    return a + b

add2 = add3

print(add3)
print(add2)

print(add3(1, 3))
r1 = add3(10, 20)
print(r1 * 100)

r2 = add2("abc", "xyz")
print(r2)

# def - 함수를 정의하는 키워드
# add - 함수 이름
# (a, b) - 매개변수(Parameter)
# : - 함수 코드블록의 시작
# return - 함수 종료와 return 뒤에 값(결과 객체)을 호출한 곳(대입문이 위치)으로 반환
```
