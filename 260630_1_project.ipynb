```python
import csv

with open("score.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    data = list(reader)

print(data)
print(type(data))
```

```python
# 1. 이상치 확인/처리
# 나이 = 평균값 반영
# 점수 n > 100 = 100 / n < 0 = 0 처리

print(type(data))
for student in data:
    kor_score = student["국어"]

    if kor_score < 0 or kor_score > 100:
        print(student)
```
