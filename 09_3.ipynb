```python
# 일반 함수 정의
def add1(a, b):
    return a + b

# 람다 함수
add2 = lambda a, b : a + b # add2에는 "lambda a, b : a + b" 람다가 반환한 함수(객체)가 대입

r1 = add1(5, 1)
print(r1)

r2 = add2(1, 3)
print(r2)
```

```python
#람다 함수를 익명함수라고 부르기도 한다
r3 = (lambda a, b : a + b)(10, 50)
print(r3)
```

```python
# 1) map() - 모든 요소를 변환

nums = [1, 2, 3, 4]

# 리스트의 각 요소(x)에 2를 곱한 결과를 반환
result = map(lambda x: x * 2, nums)

print(result) # <map object at 0x0000020E79903700>
print(list(result)) #[2, 4, 6, 8]
```

```python
# 2) filter() - 조건에 맞는 요소만 선택

nums = [1, 2, 3, 4, 5]

# 짝수인 요소만 선택
result = list(filter(lambda x: x % 2 == 0, nums))

print(result)
```

```python
# 3) sorted() - 정렬 기준 지정

names = ["banana", "apple", "kiwi"]

# 문자열의 길이를 기준으로 오름차순 정렬
result = sorted(names, key=lambda x: len(x), reverse = False) # 6, 5, 4 #기본값
result = sorted(names, key=lambda x: len(x))
print(result)

result = sorted(names, key=lambda x: len(x), reverse = True) # 4, 5, 6
print(result)

#reverse = False: 기본값으로 생략 가능
```

```python
students = [
    {"이름": "홍길동", "나이": 20, "점수": 90},
    {"이름": "이순신", "나이": 18, "점수": 85},
    {"이름": "김유신", "나이": 22, "점수": 95}
]

# "나이" 값을 기준으로 오름차순 정렬
result = sorted(students, key=lambda x: x["나이"])
print(result)

# x : 리스트의 각 딕셔너리
# x["나이"] : 각 딕셔너리의 "나이" 값을 꺼냄
# sorted()는 이 값을 기준으로 정렬한다.

# "점수"를 기준으로 내림차순 정렬
result = sorted(students, key=lambda x: x["점수"], reverse=True)
print(result)

# "이름"을 기준으로 오름차순 정렬
result = sorted(students, key=lambda x: x["이름"])
print(result)

```

```python
# 여러 키를 기준으로 정렬도 가능하다

students = [
    {"이름": "홍길동", "반": 2, "점수": 90},
    {"이름": "이순신", "반": 1, "점수": 90},
    {"이름": "김유신", "반": 1, "점수": 95}
]

# 1순위: 반, 2순위: 점수
result = sorted(students, key=lambda x: (x["반"], x["점수"]))
print(result)
```

```python
# 4) min() - 기준에 따라 최소값 찾기

names = ["banana", "apple", "kiwi"]

# 문자열 길이가 가장 짧은 요소 반환
result = min(names, key=lambda x: len(x))

print(result) # kiwi
```

```python
# 5) max() - 기준에 따라 최대값 찾기

names = ["banana", "apple", "kiwi"]

# 문자열 길이가 가장 긴 요소 반환
result = max(names, key=lambda x: len(x))

print(result) # banana
```

```python
# 6) any() - 하나라도 조건을 만족하는지 확인

nums = [1, 3, 5, 8]

# 각 요소가 짝수인지(True/False)로 변환
# 하나라도 True이면 True 반환
result = any(map(lambda x: x % 2 == 0, nums))

print(result) # True
```

```python
nums = [1, 3, 5, 8]

r1 = map(lambda x: x % 2 == 0, nums)
print("r1: ", r1)
r2 = list(r1)
print("r2: ", r2)

result = any(map(lambda x: x % 2 == 0, nums))
print(result)
```

```python
# 7) all() - 모두 조건을 만족하는지 확인

nums = [2, 4, 6, 8]

# 각 요소가 짝수인지(True/False)로 변환
# 모두 True이면 True 반환
result = all(map(lambda x: x % 2 == 0, nums))

print(result) #True
```
